```python
# ============================================================
# AI-Powered Healthcare Diagnosis Assistant
# Notebook 02: Data Preprocessing
# Step 1: Import Required Libraries
# ============================================================

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

print("Libraries imported successfully.")
```

```python
dataset_path = "/content/drive/MyDrive/AI-Powered-Healthcare-Diagnosis-Assistant/Dataset/Final_Augmented_dataset_Diseases_and_Symptoms.csv"

df = pd.read_csv(dataset_path)

print("Dataset Loaded Successfully")
print(df.shape)
```

```python
# ============================================================
# Step 5: Create Original Dataset Backup
# ============================================================

original_df = df.copy()

print("Backup created successfully.")
print("Original Dataset Shape:", original_df.shape)
```

```python
# ============================================================
# Step 6: Compare Different Rare Disease Thresholds
# ============================================================

disease_counts = original_df["diseases"].value_counts()

thresholds = [5, 10, 20]

comparison = []

for threshold in thresholds:
    valid = disease_counts[disease_counts >= threshold].index
    filtered = original_df[original_df["diseases"].isin(valid)]

    comparison.append({
        "Threshold": threshold,
        "Remaining Diseases": filtered["diseases"].nunique(),
        "Remaining Rows": filtered.shape[0],
        "Rows Removed": original_df.shape[0] - filtered.shape[0]
    })

comparison_df = pd.DataFrame(comparison)

comparison_df
```

```python
# ============================================================
# Step 7: Preview Dataset After Threshold = 10
# ============================================================

threshold = 10

disease_counts = original_df["diseases"].value_counts()

valid_diseases = disease_counts[disease_counts >= threshold].index

preview_df = original_df[original_df["diseases"].isin(valid_diseases)]

print("Preview Shape:", preview_df.shape)
print("Remaining Diseases:", preview_df["diseases"].nunique())

preview_df["diseases"].value_counts().describe()
```

```python
# ============================================================
# Step 8: Apply Rare Disease Filtering
# ============================================================

MIN_SAMPLES = 10

disease_counts = original_df["diseases"].value_counts()

valid_diseases = disease_counts[disease_counts >= MIN_SAMPLES].index

df = original_df[original_df["diseases"].isin(valid_diseases)].copy()

print("=" * 50)
print("Rare Disease Filtering Applied")
print("=" * 50)
print(f"Minimum Samples Required : {MIN_SAMPLES}")
print(f"Dataset Shape            : {df.shape}")
print(f"Remaining Diseases       : {df['diseases'].nunique()}")
```

```python
# ============================================================
# Step 9: Verify Dataset After Rare Disease Filtering
# ============================================================

print("=" * 60)
print("DATASET SUMMARY AFTER FILTERING")
print("=" * 60)

print(f"Rows                : {df.shape[0]:,}")
print(f"Columns             : {df.shape[1]}")
print(f"Remaining Diseases  : {df['diseases'].nunique()}")
print(f"Missing Values      : {df.isnull().sum().sum()}")
print(f"Duplicate Rows      : {df.duplicated().sum()}")

print("=" * 60)
```

## Observation

- Diseases having fewer than **10 samples** were removed.
- Only **433 records** were removed from the original dataset.
- The dataset still contains **246,512 records** and **677 disease classes**.
- No missing values were introduced after filtering.
- The dataset retains more than **99.8%** of the original data while improving class reliability for machine learning.

```python
# ============================================================
# Step 11: Separate Features and Target
# ============================================================

X = df.drop("diseases", axis=1)
y = df["diseases"]

print("=" * 50)
print("FEATURES AND TARGET CREATED")
print("=" * 50)
print("Feature Shape :", X.shape)
print("Target Shape  :", y.shape)
```

```python
# ============================================================
# Step 12: Label Encoding
# ============================================================

label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

print("=" * 50)
print("LABEL ENCODING COMPLETED")
print("=" * 50)
print("Total Classes :", len(label_encoder.classes_))
```

```python
# ============================================================
# Step 13: Train-Test Split
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded
)

print("=" * 50)
print("TRAIN TEST SPLIT COMPLETED")
print("=" * 50)

print("Training Set :", X_train.shape)
print("Testing Set  :", X_test.shape)
```

```python
print("=" * 50)
print("TRAINING LABELS :", len(set(y_train)))
print("TEST LABELS     :", len(set(y_test)))
print("=" * 50)
```

```python
# ============================================================
# Step 15: Save Processed Dataset
# ============================================================

processed_path = "/content/drive/MyDrive/AI-Powered-Healthcare-Diagnosis-Assistant/Processed_Data"

X_train.to_csv(f"{processed_path}/X_train.csv", index=False)
X_test.to_csv(f"{processed_path}/X_test.csv", index=False)

pd.DataFrame(y_train, columns=["Disease"]).to_csv(
    f"{processed_path}/y_train.csv", index=False
)

pd.DataFrame(y_test, columns=["Disease"]).to_csv(
    f"{processed_path}/y_test.csv", index=False
)

print("Processed datasets saved successfully.")
```
