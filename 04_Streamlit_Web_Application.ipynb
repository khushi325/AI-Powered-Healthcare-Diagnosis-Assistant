# Notebook 04: Streamlit Web Application

## Objective

The objective of this notebook is to build a professional Streamlit-based web application for the AI-Powered Healthcare Diagnosis Assistant.

The trained Logistic Regression model and Label Encoder developed in Notebook 03 are loaded into the application to perform real-time multiclass disease prediction based on user-selected symptoms.

The application provides an interactive user interface where users can select symptoms, receive disease predictions, view confidence scores, and access healthcare recommendations.

## Features

The Streamlit application includes the following features:

- Interactive Healthcare Dashboard
- Symptom Selection Interface
- AI-Based Disease Prediction
- Confidence Score Display
- Top Predicted Diseases
- Health Recommendations
- Medical Disclaimer
- Clean and Responsive User Interface

## Input

- Saved Logistic Regression Model
- Saved Label Encoder
- User Selected Symptoms

## Output

- Predicted Disease
- Prediction Confidence
- Healthcare Recommendations
- AI-Assisted Diagnosis Result

## Outcome

This notebook prepares the complete Streamlit application that will be deployed as the front-end interface of the AI-Powered Healthcare Diagnosis Assistant.

# Step 1: Import Required Libraries

In this step, all the required Python libraries for building the Streamlit application are imported.

These libraries are used for:

- Building the Streamlit user interface
- Loading the trained machine learning model
- Loading the Label Encoder
- Working with NumPy arrays
- Handling data using Pandas

# Step 2: Create the Application Header

A professional header is displayed at the top of the application.

The header introduces the project, explains its purpose, and provides users with a brief overview of the AI-powered disease prediction system.

# Step 3: Load Trained Machine Learning Model

In this step, the trained Logistic Regression model and the Label Encoder saved in Notebook 03 are loaded into the Streamlit application.

These files are required to perform real-time disease prediction based on the symptoms selected by the user.

The application also checks whether the required files are available before loading them to avoid runtime errors.

# Step 4: Load Feature Names

The trained machine learning model expects the same input features that were used during training.

Instead of manually creating the symptom list, the application loads the saved feature names generated during the preprocessing stage.

This ensures consistency between the training pipeline and the deployed application while reducing the possibility of input mismatch.

# Step 5: Create Sidebar Navigation

A professional sidebar is added to improve the user experience and navigation of the application.

The sidebar allows users to switch between different sections of the AI-Powered Healthcare Diagnosis Assistant.

Initially, the application includes the following pages:

- Home
- Disease Prediction
- About Project

The selected page is used to control the content displayed in the main application window.

# Step 6: Symptom Selection Interface

This step creates an interactive symptom selection interface using Streamlit.

Users can select one or more symptoms from the list of feature names loaded during the preprocessing stage.

The selected symptoms are displayed on the screen and will be used in the next step to generate disease predictions using the trained machine learning model.

# Step 6: Generate Disease Prediction

In this step, the selected symptoms are converted into a binary feature vector that matches the format used during model training.

Each selected symptom is assigned a value of **1**, while all remaining symptoms are assigned **0**.

The generated feature vector is passed to the trained Logistic Regression model to predict the most probable disease.
